### Defining functions

Functions take input, perform operations on this input and return the resulting value as output. To make your own function in python, you start with `def` and then the function name with parentheses that contain the input. In the example below, the function `y` takes as input `x` and performs a multiplication on `x` with variables `m` and `b`. 

```python
def y(x):
    m = 5
    b = 2
    return m*x+b
```

```python
y(3)
```

### Lists and arrays

Lists and arrays are data structures in python that allow indexing and iterating. A list is a list of data elements. Arrays are more suitable for numerical operations and are part of the numpy package. 

First create a list: 

```python
MyList = [1,2,3]
print(MyList)
```

Then, convert it to an array: 

```python
import numpy as np
MyArray = np.array(MyList)
print(MyArray)
```

You can convert it back to a list: 

```python
MyListFromArray = list(MyArray)
print(MyListFromArray)
```

If an array is inserted as a parameter within a function, it will be evaluated at each index of the array. This will not work when inserting a list. For a list, each entry would need to be evaluated in an iterative manner, for example by using a `for` loop.

To conveniently create an array of x values to be evaluated within a function we can use the numpy function `arange`. `np.arange(start, stop, step)` creates an array that begins at `start` and goes up in steps of `step`. It stops just before `stop`: the stop value itself is not included. For example, to create an array called `x_values` from 1 to 10 with a spacing of 1, use 11 as the stop value: 

```python
x_values = np.arange(1,11,1)
print(x_values)
```

Now we can evaluate the function `y` defined earlier

```python
print(y(x_values))
```

Indexing of lists and arrays in python is straightforward. Python starts counting from 0, meaning that the first index starts with 0, the second with 1 and so on. To get the second value of an array, the syntax is to first write the name of the array and then in square brackets the index: 

```python
testlist = [19,47,36]
testlist[1]
```

And for an array:

```python
testarray = np.array(testlist)
testarray[1]
```

A range of indices can be selected by taking a slice from index1 (2 in the example) to index2 (7 in the example):

```python
testarray=np.arange(1,11,1)
testarray[2:7]
```

### Plotting with matplotlib
Plots of data are usually much more insightful than tables, lists or arrays. Here we use the package matplotlib to make plots. 

First import matplotlib and plot the function `y` as defined earlier: 

```python
import matplotlib.pyplot as plt 
plt.plot(x_values,y(x_values))
plt.xlabel('x')
plt.ylabel('y(x)')
plt.show()#needed to display plot
```
