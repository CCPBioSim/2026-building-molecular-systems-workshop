**Introduction**
----------------

The package provides a *Python library* (APIs) and *Python/Bash workflows*
for automating highly parallel simulation workflows to facilitate preparation and
running atomistic and/or coarse-grain molecular simulations with the focus on
condensed soft matter systems. The package also includes tools for post-simulation
analyses.

**DOI:10.5281/zenodo.19074587** (all versions)  - https://zenodo.org/records/19074587

Shapespyer repository: https://gitlab.com/simnavi/shapespyer/

Shapespyer documentation: https://simnavi.gitlab.io/shapespyer.io/

**Motivation**
--------------

In the domain of soft matter and biomolecular simulation, the task of preparing
complex molecular structures of interest is both challenging and tedious. The naive
approach using an initial configuration with randomised positions of solute molecules
is most often fruitless due to the complexity of the (unknown) pathways that lead
to self-assembly of any particular shape of a molecular aggregate. Unfortunately,
pre-equilibration into and further relaxation of a specific molecular complex
can easily require simulation times way beyond what is reachable in a practical
simulation scenario. Therefore, *it is vital to have efficient tools facilitating
the generation and initial equilibration of pre-assembled molecular complexes with
predetermined composition and structure.* This is precisely the purpose of the
**Shapespyer** package.

Started in April 2020, the Shapespyer project is a collaboration between 
**SCD (The Shapespyer Devz team: Andrey Brukhno, Valeria Losasso, Michael Seaton)**, 
**ISIS (James Doutch, Arwel Hughes)** and **Diamond (Tim Snow)**, initially funded 
by and developed under the umbrella of the **Ada Lovelace Centre (ALC)**, aiming 
to support and facilitate interdisciplinary research by bridging between theoretical 
analysis, computer simulation and experimental studies.

**Objectives**
--------------

The primary goal of the project is to equip Small Angle Scattering (SAS) and Neutron 
Reflectometry (NR) experimentalists at ISIS and Diamond facilities in RAL with a set 
of simulation and analytical protocols allowing for the verification of hypotheses and 
theoretical models of complex molecular nanoaggregates via direct correlation of SAS and 
NR experiments to detailed computer simulations.

In turn, it is anticipated that this would facilitate the automation of common, routine,
tasks associated with the creation, simulation and analysis of complex multicomponent
aggregates, which would both simplify and standardise the use of simulations in analyses
of SAS and NR experiments. With this in mind, the Shapespyer software package has been
developed as an experimentalist-oriented Python framework providing a straightforward
workflow to facilitate the analysis of self-assembled nanostructures, ubiquitous in
soft condensed matter, and specifically those with biomolecular applications.

## **Using Jupyter notebooks**

A Jupyter notebook, when it is open in a browser, is organised in ***cells***, of which there are two types: **text** and **code**. Both types of cells can be entered for editing with a *mouse click inside the cell*. The cell that is currently in focus is highlighted with a thin vertical blue stripe on the left.

* Progress through cells by pressing `Shift` and `Enter` together - this key combination executes the cell in focus. **Note** that text cells do not produce any output but may change their mode from **Markdown editing** to **Formatted preview**.

* If you did you notice any change upon pressing `Shift`+`Enter` combination, scroll down to the next cell.

* If you see `[*]` on the left side of an executable (code) cell, it is still being executed - **wait until `*` is replaced by a number!**

* **Maintain the order of execution!**

  - DO NOT try to execute code cells in the order different from their order of appearance in the notebook! - Some cells may not execute correctly and can produce errors, due to their inputs missing or not found due to the focus being on a wrong folder (du to jumping between folders previously).

  - DO NOT switch the focus from nor execute any cell if any previous cell is still being executed, i.e. shows [*]!
